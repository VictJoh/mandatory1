## 1.2.3

$u = e^{\imath(k_x x + k_y y - \omega t)}$ gives

$$\frac{\partial^2 u}{\partial t^2} = -\omega^2 u, \quad \frac{\partial^2 u}{\partial x^2} = -k_x^2 u, \quad \frac{\partial^2 u}{\partial y^2} = -k_y^2 u$$

Inserted into $u_{tt} = c^2(u_{xx} + u_{yy})$:

$$-\omega^2 u = -c^2(k_x^2 + k_y^2) u$$

which holds since $\omega = c\sqrt{k_x^2 + k_y^2}$.

## 1.2.4

Insert $u^n_{ij} = e^{\imath(kh(i+j) - \tilde\omega n\Delta t)}$ into the discretized equation. In time:

$$u^{n+1}_{ij} - 2u^n_{ij} + u^{n-1}_{ij} = u^n_{ij}\left(e^{-\imath\tilde\omega\Delta t} - 2 + e^{\imath\tilde\omega\Delta t}\right) = u^n_{ij}\left(2\cos(\tilde\omega\Delta t) - 2\right) = -4\sin^2\left(\frac{\tilde\omega\Delta t}{2}\right)u^n_{ij}$$

Same thing in $x$ and $y$:

$$u^n_{i+1,j} - 2u^n_{ij} + u^n_{i-1,j} = -4\sin^2\left(\frac{kh}{2}\right)u^n_{ij}$$

and the same for $j$. So dividing by $u^n_{ij}$ we get

$$-\frac{4}{\Delta t^2}\sin^2\left(\frac{\tilde\omega\Delta t}{2}\right) = -\frac{8c^2}{h^2}\sin^2\left(\frac{kh}{2}\right)$$

$$\sin^2\left(\frac{\tilde\omega\Delta t}{2}\right) = 2C^2\sin^2\left(\frac{kh}{2}\right)$$

with $C = c\Delta t/h$. For $C = 1/\sqrt{2}$ we have $2C^2 = 1$ so

$$\frac{\tilde\omega\Delta t}{2} = \frac{kh}{2} \quad\Rightarrow\quad \tilde\omega = \frac{kh}{\Delta t} = \frac{kc}{C} = \sqrt{2}ck = c\sqrt{k^2 + k^2} = \omega$$
